```python
import re
def calculate_paragraph_sum():
    print("--- NUMBER SUM CALCULATOR ---")
    text = input("Please enter your paragraph:\n")
    numbers = re.findall(r'\d+', text)
    total_sum = sum(int(num) for num in numbers)
    print("\n--- Results ---")
    print(f"Numbers found: {', '.join(numbers) if numbers else 'None'}")
    print(f"Total Sum: {total_sum}")

if __name__ == "__main__":
    calculate_paragraph_sum()
```
