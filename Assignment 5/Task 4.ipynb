```python
import re
def redact_phone_numbers(text):
    pattern = r'(\+84\d+|\b\d{10}\b)'
    return re.sub(pattern, "[REDACTED]", text)
print("--- DATA PRIVACY TOOL: PHONE REDACTOR ---")
user_input = input("Please enter the text you want to process:\n")
if user_input.strip():
    redacted_result = redact_phone_numbers(user_input)
    print("\n--- Redacted Output ---")
    print(redacted_result)
else:
    print("Error: No text was provided.")
```
