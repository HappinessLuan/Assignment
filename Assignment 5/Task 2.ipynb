```python
import re
def is_valid_hex(color_code):
    pattern = r'^#[0-9A-Fa-f]{6}$'
    return bool(re.fullmatch(pattern, color_code))
print("--- WEB HEX COLOR VALIDATOR ---")
user_input = input("Enter a hex color (e.g., #FFA500): ").strip()
if is_valid_hex(user_input):
    print(f"✅ '{user_input}' is a valid hexadecimal color!")
else:
    print(f"❌ '{user_input}' is NOT a valid hexadecimal color.")
    print("Hint: It must start with '#' followed by exactly 6 hex characters (0-9, A-F).")
```
