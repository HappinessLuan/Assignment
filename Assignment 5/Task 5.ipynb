```python
import random
def approximate_pi():
    try:
        total_points = int(input("How many random points should I generate? "))
    except ValueError:
        print("Please enter a valid whole number.")
        return
    points_inside_circle = 0
    for _ in range(total_points):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        if x**2 + y**2 < 1:
            points_inside_circle += 1
    pi_approximation = 4 * points_inside_circle / total_points
    print(f"\nAfter scattering {total_points:,} points:")
    print(f"Points inside the circle: {points_inside_circle:,}")
    print(f"Approximated value of π: {pi_approximation}")

if __name__ == "__main__":
    approximate_pi()
```
