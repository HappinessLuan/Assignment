```python
import re
def is_valid_course_code(code):
    pattern = r'^[A-Z]{2,3}\d{3}$'
    if re.fullmatch(pattern, code):
        return True
    return False
print("--- UNIVERSITY COURSE CODE VALIDATOR ---")
user_input = input("Enter a course code (e.g., TEC001): ").strip()
if is_valid_course_code(user_input):
    print(f"✅ Success! '{user_input}' is a valid course code.")
else:
    print(f"❌ Invalid! '{user_input}' does not match the format (2-3 Uppercase Letters + 3 Digits).")
```
