```python
numbers = []
while True:
    user_input = input("Enter a number (or press Enter to quit): ")
    if user_input == "":
        break
    try:
        numbers.append(float(user_input))
    except ValueError:
        print("That's not a number! Try again.")
numbers.sort(reverse=True)
print("\nThe five greatest numbers are:")
print(numbers[:5])
```
