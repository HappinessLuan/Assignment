```python
def analyze_text(text):
    words = text.lower().split()
    total_words = len(words)
    if total_words == 0:
        print("The text is empty! Please try again with some content.")
        return
    counts = {}
    for word in words:
        counts[word] = counts.get(word, 0) + 1
    sorted_words = sorted(counts.items(), key=lambda item: item[1], reverse=True)
    top_5 = dict(sorted_words[:5])
    sum_top_5 = sum(top_5.values())
    proportion = (sum_top_5 / total_words) * 100
    print("\n--- Analysis Results ---")
    print(f"Top 5 most common words: {top_5}")
    print(f"Total number of words: {total_words}")
    print(f"Proportion of the top 5 words: {sum_top_5} / {total_words} = {proportion:.2f}%")
print("Word Frequency Analyzer")
user_text = input("Please enter the text you want to analyze:\n")
analyze_text(user_text)
```
