```python
def remove_odd_numbers(original_list):
    even_list = [num for num in original_list if num % 2 == 0]
    return even_list
print("Integer Filter: Keep Even Numbers Only")
user_input = input("Enter a series of integers separated by spaces (e.g., 1 5 12 8 3): ")
try:
    my_numbers = [int(x) for x in user_input.split()]
    if not my_numbers:
        print("You didn't enter any numbers!")
    else:
        filtered_numbers = remove_odd_numbers(my_numbers)
        print("\n--- Results ---")
        print(f"Original list: {my_numbers}")
        print(f"Filtered list (Even only): {filtered_numbers}")
except ValueError:
    print("Error: Please make sure to enter only whole numbers separated by spaces.")
```
