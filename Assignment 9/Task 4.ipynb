```python
filename = input("Enter the name of the file to be encrypted: ")
shift_positions = int(input("Enter the number of positions to move: "))
direction = input("Enter the direction of movement (left/right): ")

if direction.lower() == 'left':
    shift = -shift_positions
elif direction.lower() == 'right':
    shift = shift_positions
else:
    print("Invalid movement direction.")
    exit()

output_filename = "cipher_" + filename

try:
    with open(filename, 'r') as infile, open(output_filename, 'w') as outfile:
        for line in infile:
            cipher_line = ""
            for char in line:
                if char.isupper():
                    x = ord(char)
                    y = (x + shift - 65) % 26 + 65
                    cipher_line += chr(y)
                elif char.islower():
                    x = ord(char)
                    y = (x + shift - 97) % 26 + 97
                    cipher_line += chr(y)
                else:
                    cipher_line += char
            outfile.write(cipher_line)
    print(f"Encryption complete. Result file: {output_filename}")
except FileNotFoundError:
    print("File not found.")
```
