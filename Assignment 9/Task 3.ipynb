```python
filename = input("Enter the name of the file containing the scores: ")
total_score = 0.0
student_count = 0

try:
    with open(filename, 'r') as file:
        for line in file:
            line = line.strip()
            if line:
                parts = line.split(',')
                if len(parts) == 2:
                    try:
                        score = float(parts[1])
                        total_score += score
                        student_count += 1
                    except ValueError:
                        pass

    if student_count > 0:
        average = total_score / student_count
        print(f"Average score: {average}")
    else:
        print("Average score: 0.0")
except FileNotFoundError:
    print("File not found.")
```
