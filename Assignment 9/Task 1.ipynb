```python
filename = input("Enter the name of the file whose lines you want to count: ")
line_count = 0

try:
    with open(filename, 'r') as file:
        for line in file:
            if line.strip():
                line_count += 1
    print(f"Total number of lines of text: {line_count}")
except FileNotFoundError:
    print("File not found")
```
