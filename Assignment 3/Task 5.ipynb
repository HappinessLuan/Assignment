```python
attempts = 0
success = False
while attempts < 5:
    username = input('Username:')
    password = input('Password:')
    if username == 'python' and password == 'rules':
        success = True
        break
    else:
        attempts += 1
        print(f'Incorrect. {5 - attempts} attempts remaining.')
print('Welcome' if success else 'Access denied')
```
