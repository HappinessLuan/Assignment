```python
def extract_middle(text):
    length = len(text)
    middle = length // 2
    if length % 2 == 0:
        return text[middle - 1 : middle + 1]
    else:
        return text[middle]

input = input('Enter a string:')
middle_character = extract_middle(input)
print('Middle character(text):', middle_character)
```
