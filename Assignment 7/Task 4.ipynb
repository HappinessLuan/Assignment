```python
import random

class Car:
    def __init__(self, registration_number, maximum_speed):
        self.registration_number = registration_number
        self.maximum_speed = maximum_speed
        self.current_speed = 0
        self.travelled_distance = 0

    def accelerate(self, speed_change):
        self.current_speed += speed_change

        if self.current_speed < 0:
            self.current_speed = 0
        elif self.current_speed > self.maximum_speed:
            self.current_speed = self.maximum_speed

    def drive(self, hours):
        self.travelled_distance += self.current_speed * hours


def main():
    # 1. Create a list of 10 cars
    cars = []
    for i in range(1, 11):
        reg_num = f"ABC-{i}"
        max_speed = random.randint(150, 200)
        cars.append(Car(reg_num, max_speed))

    # 2. Run the race
    race_finished = False
    hours_passed = 0

    while not race_finished:
        hours_passed += 1
        for car in cars:
            # Random speed change between -10 and +15 km/h
            speed_change = random.randint(-10, 15)
            car.accelerate(speed_change)

            # Drive for 1 hour
            car.drive(1)

            # Check if this car has reached 10,000 km
            if car.travelled_distance >= 10000:
                race_finished = True

    # 3. Print the results in a formatted table
    print(f"The race finished after {hours_passed} hours!\n")

    # Table header
    print(f"{'Registration':<15} | {'Max Speed (km/h)':<18} | {'Current Speed (km/h)':<22} | {'Distance (km)':<15}")
    print("-" * 78)

    # Table rows
    for car in cars:
        print(f"{car.registration_number:<15} | {car.maximum_speed:<18} | {car.current_speed:<22} | {car.travelled_distance:<15}")


if __name__ == "__main__":
    main()
```
