```python
class Car:
    def __init__(self, registration_number, maximum_speed):
        self.registration_number = registration_number
        self.maximum_speed = maximum_speed
        self.current_speed = 0
        self.travelled_distance = 0

    def accelerate(self, speed_change):
        self.current_speed += speed_change

        if self.current_speed < 0:
            self.current_speed = 0
        elif self.current_speed > self.maximum_speed:
            self.current_speed = self.maximum_speed

    def drive(self, hours):
        self.travelled_distance += self.current_speed * hours


def main():
    print("=== CREATE NEW CAR ===")
    # Get input data from the user
    reg_num = input("Enter registration number (e.g., ABC-123): ")
    max_spd = int(input("Enter maximum speed of the car (km/h): "))

    # Create a Car object with the entered information
    my_car = Car(reg_num, max_spd)

    print("\n--- Newly created car info ---")
    print(f"Registration number: {my_car.registration_number}")
    print(f"Maximum speed: {my_car.maximum_speed} km/h")
    print(f"Current speed: {my_car.current_speed} km/h")
    print(f"Travelled distance: {my_car.travelled_distance} km\n")

    print("=== TEST ACCELERATION / BRAKING ===")
    # Allow the user to input the speed change
    speed_change_1 = int(input("Enter the first speed change (km/h): "))
    my_car.accelerate(speed_change_1)
    print(f"-> Current speed: {my_car.current_speed} km/h")

    speed_change_2 = int(input("Enter the second speed change (km/h, enter a negative number to brake): "))
    my_car.accelerate(speed_change_2)
    print(f"-> Current speed: {my_car.current_speed} km/h\n")

    print("=== TEST DRIVING ===")
    # Allow the user to input drive time (can be a decimal)
    drive_time = float(input("Enter driving time at the current speed (hours): "))
    my_car.drive(drive_time)

    print("\n=== FINAL RESULT ===")
    print(f"Total distance travelled by the car: {my_car.travelled_distance} km")


if __name__ == "__main__":
    main()
```
