```python
def remove_uneven(numbers):
    """
    Takes a list of integers and returns a new list containing
    only the even numbers.
    """
    even_numbers = []

    for num in numbers:
        if num % 2 == 0:
            even_numbers.append(num)
    return even_numbers
def main():
    original_list = [12, 209, 31221, 421, 52, 61, 7, 88, 9, 40, 12, 13]
    even_only_list = remove_uneven(original_list)
    print(f"Original list: {original_list}")
    print(f"Even numbers only: {even_only_list}")

if __name__ == "__main__":
    main()
```
