```python
def calculate_sum(numbers):
    """
    Takes a list of integers and returns the total sum.
    """
    total = 0
    for num in numbers:
        total += num
    return total
def main():
    my_numbers = [11, 25, 5, 40, 20]
    result = calculate_sum(my_numbers)
    print(f"The list of numbers is: {my_numbers}")
    print(f"The sum of all numbers in the list is: {result}")

if __name__ == "__main__":
    main()
```
