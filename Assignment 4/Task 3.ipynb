```python
def main():
    cities = []
    print("Please enter the names of five cities:")
    for i in range(5):
        city_name = input(f"Enter city name {i + 1}: ")
        cities.append(city_name)
    print("\n--- Your List of Cities ---")
    for city in cities:
        print(city)

if __name__ == "__main__":
    main()
```
